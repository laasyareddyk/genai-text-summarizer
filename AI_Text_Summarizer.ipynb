```python
import google.generativeai as genai
```

```python
API_KEY = input("Enter your Gemini API key: ")

genai.configure(api_key=API_KEY)
```

```python
model = genai.GenerativeModel("gemini-3.6-flash")

print("Gemini model initialized successfully!")
```

```python
text = input("Paste the text you want to summarize:\n\n")

if not text.strip():
    print("❌ No text entered!")
else:
    print("\n✅ Text received!")
    print("Number of words:", len(text.split()))
```

```python
prompt = f"""
You are an AI text summarization assistant.

Summarize the following text.

Give the output in this format:

SUMMARY:
Write a clear and concise summary.

KEY POINTS:
- Point 1
- Point 2
- Point 3
- Point 4
- Point 5

KEYWORDS:
List the important keywords.

MAIN TOPIC:
State the main topic in one sentence.

TEXT:
{text}
"""

response = model.generate_content(prompt)

print(response.text)
```

```python
original_words = len(text.split())
output_words = len(response.text.split())

print("\n" + "="*50)
print("TEXT STATISTICS")
print("="*50)

print("Original text words :", original_words)
print("AI output words     :", output_words)

reduction = (1 - output_words / original_words) * 100

print("Word reduction      :", round(reduction, 2), "%")
```
